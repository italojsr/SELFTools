```python
from classes import Rele,Arquivo,ReleFTP
from ftplib import FTP
import pandas as pd 
import shutil
import os   
import numpy as np
```

```python
def autentica_ftp(SE,EQP,IP):
    # Tenta autenticacao com o ip da rede
        try:
            rl = ReleFTP(IP,"2AC","TAIL",SE,EQP)
            return rl
        except:
            # Nenhuma senha deu certo, será adicionada uma linha com o ip pingado e essa observação
            return None
```
