```python
from classes import Rele,Arquivo,ReleFTP
from ftplib import FTP
import pandas as pd 
import shutil
import os   
import numpy as np
```

```python
def autentica_ftp(SE,EQP,IP):
    # Tenta autenticacao com o ip da rede
        try:
            rl = ReleFTP(IP,"2AC","TAIL",SE,EQP)
            return rl
        except:
            # Nenhuma senha deu certo, será adicionada uma linha com o ip pingado e essa observação
            return None
```

```python
linhas_puladas1,linhas_puladas2 = [0,0]
with open('LAB_751_105_SET_1__atualizado.TXT', "r") as arquivo:
    for linha in arquivo.readlines():
        linhas_puladas1+=1
        if((linha.find("[") != -1) and (linha != "[INFO]\n")): 
            break
with open('LAB_751_105_105_SET_1__backup.TXT', "r") as arquivo:
    for linha in arquivo.readlines():
        linhas_puladas2+=1
        if((linha.find("[") != -1) and (linha != "[INFO]\n")):
            break
# Lê os arquivos pulando as primeiras linhas
arq_backup = pd.read_csv('LAB_751_105_105_SET_1__backup.TXT',skiprows = linhas_puladas1,sep = ',')
arq_atualizado = pd.read_csv('LAB_751_105_SET_1__atualizado.TXT',skiprows = linhas_puladas2,sep = ',')
# Compara o arquivo baixado com o arquivo atualizado e retorna um DataFrame com diferenças
comparacao_df = arq_backup.compare(arq_atualizado)
```

#### Fluxo Principal - Enviar arquivo com até 2 alterações

```python
# Conecta a algum equipamento informando nomes de SE,EQP,IP
SE = 'LAB'
EQP = 'ALTA_T1'
IP = '10.10.10.10'
rl = autentica_ftp(SE,EQP,IP)

# Baixa o arquivo desejado
ARQ = 'SET_1.TXT'
rl.baixar_arquivos_ftp(ARQ)

# Abre o arquivo desejado
ARQ = ARQ.split(".")[0]
atualizado = f'{SE}_{EQP}_{ARQ}__atualizado.TXT'
arq = rl.abrir_arquivo(atualizado)

# Faz backup 
backup = arq.fazer_backup()

```

Verifica o parâmetro a ser alterado

```python
arq.verificar_parametro('TR')
```

Atualiza o parâmetro

```python
arq.atualizar_parametro('TR','RB20 # 51P1T OR 51G1T OR 51N1T OR 50P1T OR 50G1T OR 50N1T OR IN407 # IN407 = GAS T2')

```

Verifica se ele foi alterado

```python
arq.verificar_parametro('TR')
```

Envia o arquivo

```python
# Envia arquivo
rl.enviar_arquivo_atualizado(backup,atualizado)
```

Restaura o backup

```python
# Restaura backup
rl.enviar_backup(backup,'SET_1.TXT')
```

Verifica se o backup foi restaurado com sucesso

```python
arq2 = rl.abrir_arquivo(f'{SE}_{EQP}_{ARQ}__baixado.TXT')
arq2.verificar_parametro('TR')
```
