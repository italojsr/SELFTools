```python
from classes import Rele,Arquivo,ReleFTP
from ftplib import FTP
import pandas as pd 
import shutil
import os   
import numpy as np
```

```python
def autentica_ftp(SE,EQP,IP):
    # Tenta autenticar com as duas senhas possíveis. Se não for autenticado retorna None
    try:
        rl = ReleFTP(IP,"2AC","EQ.PR0T@eqtl",SE,EQP)
        return rl
    except:
        try:
            rl = ReleFTP(IP,"2AC","TAIL",SE,EQP)
            return rl
        except:
            # Nenhuma senha deu certo, será adicionada uma linha com o ip pingado e essa observação
            return None
```
